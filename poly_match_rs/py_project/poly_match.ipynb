# Poly Match (Python)

Ce notebook mesure et compare les versions `poly_match_v0` → `poly_match_v4`.

Prérequis:
- Avoir activé la venv
- Avoir compilé le module Rust depuis `../` avec `maturin develop --release`
- Redémarrer le kernel après recompilation


```python
# Liste des modules disponibles
modules = [
    "poly_match_v0",
    "poly_match_v1",
    "poly_match_v2",
    "poly_match_v3",
    "poly_match_v4",
]
modules
```

```python
import importlib, time

def measure(module_name: str, n_iter: int = 10):
    m = importlib.import_module(module_name)
    t0 = time.perf_counter()
    for _ in range(n_iter):
        m.run_once()  # suppose une API simple; sinon, adapter
    dt = (time.perf_counter() - t0) * 1000 / n_iter
    return dt

for name in modules:
    try:
        avg_ms = measure(name)
        print(f"{name:>16}  avg: {avg_ms:8.2f} ms/iter")
    except Exception as e:
        print(f"{name:>16}  ERROR: {e}")
```
